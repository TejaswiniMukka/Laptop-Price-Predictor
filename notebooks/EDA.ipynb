# Laptop Price Predictor - Exploratory Data Analysis (EDA)

This notebook performs an in-depth exploratory analysis of the laptop price dataset to understand the relationships, distributions, and patterns in the data.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Set plotly template to professional dark/light theme
pio.templates.default = "plotly_white"

# Load dataset
df = pd.read_csv('../data/laptop_data.csv', encoding='latin-1')
print(f"Dataset shape: {df.shape}")
df.head()
```

### Chart 1: Dataset Overview
We inspect the column data types and check for any immediate format issues.

```python
df.info()
```

### Missing Value Analysis
We count the number of missing values in each column to identify if imputation is required.

```python
print("Missing values:")
print(df.isnull().sum())
```

### Duplicate Detection
We check if there are any duplicate records in the dataset.

```python
print(f"Number of duplicate rows: {df.duplicated().sum()}")
```

### Statistical Summary
We print descriptive statistics for numerical and categorical columns.

```python
df.describe(include='all')
```

### Chart 2: Target Variable (Price_euros) Distribution
We visualize the distribution of laptop prices. Typically, price distributions are right-skewed, which might suggest a log-transformation during modeling (or robust models like Random Forest).

```python
fig_price = px.histogram(df, x='Price_euros', marginal='box', 
                         title='Distribution of Laptop Prices (in Euros)',
                         color_discrete_sequence=['#636EFA'])
fig_price.show()
```

### Chart 3: Laptop Brand (Company) Distribution
We examine which manufacturers dominate the dataset. This helps us see if we have enough sample sizes for all brands.

```python
brand_counts = df['Company'].value_counts().reset_index()
brand_counts.columns = ['Company', 'Count']
fig_brand = px.bar(brand_counts, x='Company', y='Count', 
                    title='Laptop Brand Distribution',
                    color='Count', color_continuous_scale='Purples')
fig_brand.show()
```

### Chart 4: Laptop Type (TypeName) Distribution
We inspect the counts of laptop types (e.g. Notebook, Ultrabook, Gaming, 2 in 1 Convertible) to see consumer categories.

```python
type_counts = df['TypeName'].value_counts().reset_index()
type_counts.columns = ['TypeName', 'Count']
fig_type = px.pie(type_counts, names='TypeName', values='Count', 
                    title='Laptop Type Distribution',
                    color_discrete_sequence=px.colors.qualitative.Pastel)
fig_type.show()
```

### Chart 5: Operating System Distribution
We look at the spread of operating systems.

```python
opsys_counts = df['OpSys'].value_counts().reset_index()
opsys_counts.columns = ['OpSys', 'Count']
fig_opsys = px.bar(opsys_counts, x='OpSys', y='Count', 
                    title='Operating System Distribution',
                    color='Count', color_continuous_scale='Viridis')
fig_opsys.show()
```

### Chart 6: RAM Distribution
Before analyzing RAM, we clean it (extracting number from "8GB").

```python
df_clean = df.copy()
df_clean['Ram_GB'] = df_clean['Ram'].str.replace('GB', '', regex=False).astype(int)
ram_counts = df_clean['Ram_GB'].value_counts().sort_index().reset_index()
ram_counts.columns = ['Ram_GB', 'Count']
fig_ram = px.bar(ram_counts, x='Ram_GB', y='Count', 
                  title='RAM Size Distribution (GB)',
                  color='Count', color_continuous_scale='Cividis')
fig_ram.update_layout(xaxis=dict(type='category'))
fig_ram.show()
```

### Chart 7: Price Distribution by Company
A boxplot showing price variations by brand. This shows that brands like Apple, Dell, and MSI target higher price ranges while Acer targets budget markets.

```python
fig_price_brand = px.box(df, x='Company', y='Price_euros', 
                         title='Price Distribution by Company', color='Company')
fig_price_brand.show()
```

### Correlation Analysis & Outlier Detection
We clean the data using our modular preprocessing pipeline to convert textual features to numeric values, enabling correlation analysis.

```python
sys.path.append('../')
from utils.preprocess import preprocess_pipeline

df_preprocessed = preprocess_pipeline(df, is_training=True)

# Select numerical columns
num_cols = df_preprocessed.select_dtypes(include=[np.number])
corr_matrix = num_cols.corr()

fig_corr = px.imshow(corr_matrix, text_auto=True, 
                      title='Correlation Matrix of Features',
                      color_continuous_scale='RdBu_r', aspect='auto')
fig_corr.show()
```

### Chart 8: Price vs. PPI (Pixels Per Inch)
We investigate how screen density relates to pricing.

```python
fig_ppi = px.scatter(df_preprocessed, x='PPI', y='Price_euros', 
                     title='Price vs. Pixels Per Inch (PPI)',
                     color='TypeName', hover_name='Company')
fig_ppi.show()
```
